# HOT CASE STUDY
### **Vaccination Dashboard**: Combine Pandas groupby + bar charts (age), line charts (cumulative), and optional Plotly maps; interpret which regions require policy attention.

####INTRODUCTION
Vaccination campaigns are essential for public health protection and disease control. Monitoring vaccination coverage across different age groups and regions enables policymakers to identify disparities and allocate resources effectively.

This case study develops a Vaccination Dashboard using data science tools to analyze vaccination distribution patterns and identify regions requiring policy attention. The dashboard integrates data aggregation, statistical reasoning, and visual interpretation to support evidence-based decision-making.

#### TECHNICAL IMPLEMENTATION

Tools Used:

Python

Pandas – Data aggregation using groupby

Matplotlib – Static bar and line charts

Plotly Express – Interactive choropleth map

NumPy – Sample dataset generation

```python

# Importing libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import plotly.express as px

# Sample Dataset

np.random.seed(42)

dates = pd.date_range(start='2025-01-01', periods=30)
regions = ['North', 'South', 'East', 'West']
age_groups = ['18-25', '26-45', '46-60', '60+']

data = []

for date in dates:
    for region in regions:
        for age in age_groups:
            vaccinated = np.random.randint(100, 500)
            data.append([date, region, age, vaccinated])

df = pd.DataFrame(data, columns=['Date', 'Region', 'Age_Group', 'Vaccinated_Count'])

print("Sample Data Preview:")
print(df.head())



# Age-wise Vaccination Analysis


age_group_data = df.groupby('Age_Group')['Vaccinated_Count'].sum().reset_index()

plt.figure()
plt.bar(age_group_data['Age_Group'], age_group_data['Vaccinated_Count'])
plt.xlabel("Age Group")
plt.ylabel("Total Vaccinated")
plt.title("Age-wise Vaccination Distribution")
plt.show()

print("\nAge-wise Vaccination Totals:")
print(age_group_data.sort_values(by='Vaccinated_Count'))


# Cumulative Vaccination Trend

daily_data = df.groupby('Date')['Vaccinated_Count'].sum().sort_index()
cumulative_data = daily_data.cumsum()

plt.figure()
plt.plot(cumulative_data.index, cumulative_data.values)
plt.xlabel("Date")
plt.ylabel("Cumulative Vaccination")
plt.title("Cumulative Vaccination Over Time")
plt.xticks(rotation=45)
plt.show()



# Region-wise Vaccination Analysis

region_data = df.groupby('Region')['Vaccinated_Count'].sum().reset_index()

plt.figure()
plt.bar(region_data['Region'], region_data['Vaccinated_Count'])
plt.xlabel("Region")
plt.ylabel("Total Vaccinated")
plt.title("Region-wise Vaccination Distribution")
plt.show()

# Calculate Percentage Share
total_vaccinated = region_data['Vaccinated_Count'].sum()
region_data['Percentage_Share'] = (
    region_data['Vaccinated_Count'] / total_vaccinated * 100
)

region_sorted = region_data.sort_values(by='Vaccinated_Count')

print("\nRegion-wise Vaccination Totals:")
print(region_sorted)



# Policy Attention Identification


lowest_region = region_sorted.iloc[0]

print("\n==============================")
print(" POLICY ATTENTION ANALYSIS ")
print("==============================")

print(f"Region requiring immediate policy attention: {lowest_region['Region']}")
print(f"Total Vaccinated: {lowest_region['Vaccinated_Count']}")
print(f"Percentage Share: {lowest_region['Percentage_Share']:.2f}%")

print("\nInterpretation:")
print(f"The {lowest_region['Region']} region has the lowest vaccination coverage.")
print("This suggests potential healthcare access or awareness gaps.")
print("Focused vaccination drives and infrastructure support are recommended.")



print("\nDashboard Analysis Complete Successfully.")
```

The lowest-performing region requires policy intervention.
